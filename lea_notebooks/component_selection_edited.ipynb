```python
#Convert tsv data from signor to sif format
import pandas as pd

df = pd.read_csv("data\\all_data_M_musculus_07_10_26.tsv", sep="\t")
# SIGNOR utilise 'ENTITYA', 'MECHANISM'/'EFFECT', et 'ENTITYB'
sif_df = df[["ENTITYA", "EFFECT", "ENTITYB"]]
#sif_df.to_csv("data\\signor_network.sif", sep="\t", index=False, header=False)
```

```python
def df_to_interaction_graph(df):

    interaction_graph = []

    for _, row in df.iterrows():
        source = str(row["ENTITYA"]).strip()
        target = str(row["ENTITYB"]).strip()
        effect = str(row["EFFECT"]).strip().lower()

        # Vérification si la chaîne contient "up-regulates" ou "down-regulates"
        if "up-regulates" in effect or "activates" in effect:
            sign = 1
        elif "down-regulates" in effect or "inhibits" in effect:
            sign = -1
        else:
            # Si aucun des termes clés n'est trouvé dans le texte de l'effet
            continue

        interaction_graph.append((source, target, {"sign": sign}))

    return interaction_graph


# --- Utilisation ---
interaction_graph = df_to_interaction_graph(sif_df)

# Aperçu du résultat
print(interaction_graph[:5])
```

```python
#Standardize the gene names in the interaction graph using the same metadata gene matrix as for the single-cell dataset
stand_signor= gns.interaction_list_standardization(interaction_graph, "../data/Mus_musculus.gene_info.202103.tsv")
print(stand_signor[:5])
```

```python
#Filter only on data that is in the scRNA seq dataset
filtered_signor = []

for source, target, attr in stand_signor:
    # On vérifie que la SOURCE ET la CIBLE sont dans le jeu de données scRNA-seq
    if source in standardized_genenames_in_singlecell and target in standardized_genenames_in_singlecell:
        filtered_signor.append((source, target, attr))

print(f"Nombre d'interactions initiales : {len(stand_signor)}")
print(f"Nombre d'interactions filtrées  : {len(filtered_signor)}")
```

```python
interaction_graph = bonesis.InfluenceGraph(filtered_signor)
```

```python
#INitialization of the research space for boolean networks
bo = bonesis.BoNesis(interaction_graph, standardized_observations)
```

```python
print(f"domain: {len(interaction_graph.nodes())} nodes, {len(interaction_graph.edges())} edges")
```

#### fixpoints

```python
s2 = bo.fixed(~bo.obs("S2"))
s4 = bo.fixed(~bo.obs("S4"))
s5 = bo.fixed(~bo.obs("S5"))
s2 != s4
s5 != s4
s2 != s5;
```

#### positive reachability

```python
~bo.obs('S1') >= ~bo.obs('S0') >= s2
~bo.obs('S0') >= ~bo.obs('S3') >= s4
~bo.obs('S3') >= s5;
```

#### negative reachability

```python
~bo.obs("S3") / s2;
```
